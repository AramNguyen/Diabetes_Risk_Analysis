```python
import pandas as pd
from pandas.api.types import is_numeric_dtype

output_path = '02_data-cleaned.csv'

df = pd.read_csv('00-diabetes-MAIN-DATASET.csv')

# delete patient_id (or the machine will be like: "oh Patient #4102? So he must have diabetes")
if 'patient_id' in df.columns:
    df = df.drop('patient_id', axis=1)

# imputation
# number: mean number (giá trị trung bình), string -> "Unknown"
for col in df.columns:
    if df[col].isnull().sum() > 0:
        if is_numeric_dtype(df[col]):
            df[col] = df[col].fillna(df[col].mean())
        else:
            df[col] = df[col].fillna("Unknown")

# ordinal encoding
risk_mapping = {'Low': 0, 'Moderate': 1, 'High': 2}
if 'diabetes_risk' in df.columns:
    df['diabetes_risk'] = df['diabetes_risk'].map(risk_mapping)

# income_bracket to int
# "Unknown" -> -1
income_mapping = {'Low': 0, 'Middle': 1, 'High': 2, 'Unknown': -1}
if 'income_bracket' in df.columns:
    df['income_bracket'] = df['income_bracket'].map(income_mapping)

# one-hot encoding
df_cleaned = pd.get_dummies(df, drop_first=True)

# output (clean data)
df_cleaned.to_csv(output_path, index=False)

print(df_cleaned[['diabetes_risk']].head(10))
```
