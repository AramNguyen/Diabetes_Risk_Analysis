```python
# Này là code cũ của ông
import pandas as pd
from pandas.api.types import is_numeric_dtype

# 1. Khai báo đường dẫn
input_path = 'diabetes_risk.csv'
output_path = 'clean_data_diabetes.csv'

# 2. Nạp dữ liệu
df = pd.read_csv(input_path)

# 3. Xóa định danh cá nhân
if 'patient_id' in df.columns:
    df.drop('patient_id', axis=1, inplace=True)

# 4. Vá lỗi dữ liệu (Imputation)
# Chỉ lấp đầy các ô trống, không làm biến đổi cấu trúc cột
for col in df.columns:
    if df[col].isnull().sum() > 0:
        if is_numeric_dtype(df[col]):
            df[col].fillna(df[col].mean(), inplace=True)
        else:
            df[col].fillna(df[col].value_counts().idxmax(), inplace=True)

# 5. Chuyển đổi biến thứ bậc (Ordinal Encoding)
# Ánh xạ thủ công cột diabetes_risk thành số nguyên 0, 5, 9
risk_mapping = {'Low': 0, 'Moderate': 5, 'High': 9}
if 'diabetes_risk' in df.columns:
    df['diabetes_risk'] = df['diabetes_risk'].map(risk_mapping)

# (Lưu ý: Nếu cột income_bracket có các nhãn Low, Middle, High mà bạn cũng muốn 
# chuyển thành 0, 5, 9 thì mở comment 2 dòng code bên dưới để chạy)
# income_mapping = {'Low': 0, 'Middle': 5, 'High': 9}
# df['income_bracket'] = df['income_bracket'].map(income_mapping)

# 6. Tạo biến giả (One-hot Encoding / Dummy)
# Hàm này sẽ chỉ băm các cột còn sót lại đang là chữ (object/string)
df_final = pd.get_dummies(df, drop_first=True)

# 7. Xuất file hoàn chỉnh
df_final.to_csv(output_path, index=False)

# 8. Báo cáo kiểm tra
print("Đã giữ lại và chuyển hóa thành công cột diabetes_risk thành 0-5-9!")
print(f"Kích thước dữ liệu cuối cùng: {df_final.shape}")
print("\n5 dòng đầu tiên của cột mục tiêu (đã ánh xạ):")
print(df_final[['diabetes_risk']].head())

# Tui để note - kh xóa cho ông đọc cái nha Phúc - ông đọc xong ông xóa cũng được 
```

```python
import pandas as pd
from pandas.api.types import is_numeric_dtype

output_path = '02_data-cleaned.csv'

df = pd.read_csv('00-diabetes-MAIN-DATASET.csv')

# delete patient_id (or the machine will be like: "oh Patient #4102? So he must have diabetes")
if 'patient_id' in df.columns:
    df = df.drop('patient_id', axis=1)

# imputation
# number: mean number (giá trị trung bình), string -> "Unknown"
for col in df.columns:
    if df[col].isnull().sum() > 0:
        if is_numeric_dtype(df[col]):
            df[col] = df[col].fillna(df[col].mean())
        else:
            df[col] = df[col].fillna("Unknown")

# ordinal encoding
risk_mapping = {'Low': 0, 'Moderate': 1, 'High': 2}
if 'diabetes_risk' in df.columns:
    df['diabetes_risk'] = df['diabetes_risk'].map(risk_mapping)

# income_bracket to int
# "Unknown" -> -1
income_mapping = {'Low': 0, 'Middle': 1, 'High': 2, 'Unknown': -1}
if 'income_bracket' in df.columns:
    df['income_bracket'] = df['income_bracket'].map(income_mapping)

# one-hot encoding
df_cleaned = pd.get_dummies(df, drop_first=True)

# output (clean data)
df_cleaned.to_csv(output_path, index=False)

print(df_cleaned[['diabetes_risk']].head(10))
```
