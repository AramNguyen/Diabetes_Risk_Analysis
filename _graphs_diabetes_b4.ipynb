```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler

df = pd.read_csv('/Users/hungdan/Documents/Projects/project_01/Diabetes_Risk_Analysis/clean_data_diabetes.csv')

features_to_scale = ['bmi', 'fasting_blood_sugar', 'age', 'waist_circumference_cm']
scaler = StandardScaler()
df_scaled = df.copy()
df_scaled[features_to_scale] = scaler.fit_transform(df[features_to_scale])


sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(2, 2, figsize=(14, 10))


sns.boxplot(x='diabetes_risk', y='bmi', data=df_scaled, ax=axes[0, 0], color='lightblue', showfliers=False)
sns.stripplot(x='diabetes_risk', y='bmi', data=df_scaled, ax=axes[0, 0], alpha=0.05, jitter=True, color='darkblue')
axes[0, 0].set_title('Rủi ro tiểu đường vs BMI (Đã chuẩn hóa Z-score)')

sns.boxplot(x='diabetes_risk', y='fasting_blood_sugar', data=df_scaled, ax=axes[0, 1], color='lightgreen', showfliers=False)
sns.stripplot(x='diabetes_risk', y='fasting_blood_sugar', data=df_scaled, ax=axes[0, 1], alpha=0.05, jitter=True, color='darkgreen')
axes[0, 1].set_title('Rủi ro tiểu đường vs Đường huyết đói (Z-score)')

sns.boxplot(x='diabetes_risk', y='age', data=df_scaled, ax=axes[1, 0], color='lightcoral', showfliers=False)
sns.stripplot(x='diabetes_risk', y='age', data=df_scaled, ax=axes[1, 0], alpha=0.05, jitter=True, color='darkred')
axes[1, 0].set_title('Rủi ro tiểu đường vs Độ tuổi (Z-score)')

sns.boxplot(x='diabetes_risk', y='waist_circumference_cm', data=df_scaled, ax=axes[1, 1], color='thistle', showfliers=False)
sns.stripplot(x='diabetes_risk', y='waist_circumference_cm', data=df_scaled, ax=axes[1, 1], alpha=0.05, jitter=True, color='purple')
axes[1, 1].set_title('Rủi ro tiểu đường vs Vòng eo (Z-score)')

plt.tight_layout()
plt.show()
```
